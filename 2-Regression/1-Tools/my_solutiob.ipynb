```python
situps_features = features[['Situps']]
target = targets['Waist']
x_train, x_test, y_train, y_test = train_test_split(features, target, test_size=0.2, random_state=42)
```

```python
model = LinearRegression()

model.fit(x_train, y_train)

model_predictions = model.predict(x_test)
```

```python
plt.scatter(x_test, y_test, color='blue', label='Actual')
plt.plot(x_test, model_predictions, color='red', linewidth=2, label='Predicted')
plt.xlabel('Situps')
plt.ylabel('Waist Circumference')
plt.title('Linear Regression: Situps vs Waist Circumference')
plt.legend()
plt.show()
```

```python
x = features
y = targets['Waist']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

developed_model = LinearRegression()

developed_model.fit(x_train, y_train)

developed_model_predictions = developed_model.predict(x_test)
```
